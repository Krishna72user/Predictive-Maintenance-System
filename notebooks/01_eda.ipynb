```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/ai4i2020.csv')
```

```python
df.head()
```

```python
df = df.iloc[:,2:9]
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
sns.countplot(x = 'Machine failure',data = df)
```

```python
fig,axs = plt.subplots(5,2,figsize=(8,20))
for i in range(5):
    for j in range(2):
        if j == 0:
            sns.histplot(df.iloc[:,i],ax = axs[i,j],kde=True)
        else:
            sns.boxplot(df.iloc[:,i],ax = axs[i,j])
fig.tight_layout()
```

```python
cor_m = df.iloc[:,1:6].corr()
sns.heatmap(cor_m,annot = True,fmt = '.2f')
plt.show()
```
